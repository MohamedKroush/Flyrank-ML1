## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

For content reviewers with limited review capacity, deciding which pages to review first, we will build a ranked scoring queue from the FlyRank content/search dataset, predicting/scoring which pages are likely to decline over the next 30 days, measured by Precision@50. A wrong call costs wasted reviewer time (false positive) or a missed real decline (false negative). A plain rule isn't enough because decline signals come from many tangled, shifting factors — staleness, visibility, position, and engagement — that interact in ways a simple if/else can't capture reliably. We will claim only observed, directional, decision-support results.

## 3. Quick look at the data (2-3 real numbers)

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_rate = df["trend_direction"].str.lower().eq("down").mean()
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
no_position_data = (df["avg_position"] == 0).sum()

print(f"Declining rate: {declining_rate:.1%}")
print(f"Stale + visible pages (review candidates): {stale_visible} of {len(df)}")
print(f"Rows with no position data (avg_position == 0): {no_position_data}")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

This work can say a page is observed as declining or flagged as stale/visible in a defined window — measured facts. It can offer directional, decision-support rankings: which pages deserve a reviewer's attention first. It cannot claim a refresh will cause recovery (no causal experiment here), and it cannot claim to have found a Google ranking factor.
